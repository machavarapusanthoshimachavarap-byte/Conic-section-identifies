```python
# Enter A, B, C, D, E, F

A = 1
B = 0
C = 4
D = 0
E = 0
F = -4

print("Your conic is:")
print(f"{A}x² + {B}xy + {C}y² + {D}x + {E}y + {F} = 0")
```

```python
import numpy as np

Q = np.array([
    [A, B/2],
    [B/2, C]
])

print(Q)
```

```python
theta = 0.5 * np.arctan2(B, A - C)

theta_degrees = np.degrees(theta)

print("Rotation angle:")
print(theta_degrees, "degrees")
```

```python
det_Q = np.linalg.det(Q)

print("Determinant of Q =", det_Q)

if det_Q > 0 and A + C > 0:
    conic_type = "Ellipse"
elif det_Q > 0 and A + C < 0:
    conic_type = "Imaginary ellipse"
elif det_Q < 0:
    conic_type = "Hyperbola"
elif abs(det_Q) < 1e-10:
    conic_type = "Parabola"
else:
    conic_type = "Other / degenerate"

print("Conic type:", conic_type)
```
